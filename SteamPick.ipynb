# SteamPick: Game Recommendation Tool

This app helps Twitch and YouTube streamers decide which Steam games to play next by comparing popularity, average playtime, and player ratings.

The app answers three questions:
1. Which games have the highest peak concurrent players?
2. Which games have the longest average playtime?
3. Which games have the highest percentage of positive reviews?


```python
import pandas as pd

url = "https://raw.githubusercontent.com/meghanthota/MIST-4600-SteamPick/main/steam_games.csv"

games = pd.read_csv(url)

print("Loaded", len(games), "games")
games.head()
```


## Question 1: Which games are the most popular?

This function finds the games with the highest peak number of concurrent players.


```python
def most_popular(games):
    popular = games.sort_values(
        "peak_ccu",
        ascending=False
    )

    return popular[["name", "peak_ccu"]].head(5)

most_popular(games)
```


## Question 2: Which games have the longest average playtime?

This function identifies games that players spend the most time playing. Longer average playtime may suggest that a game can keep players engaged over time.


```python
def longest_playtime(games):
    playtime = games.sort_values(
        "avg_playtime_hours",
        ascending=False
    )

    return playtime[["name", "avg_playtime_hours"]].head(5)

longest_playtime(games)
```


## Question 3: Which games have the highest positive review percentages and how do ratings relate to popularity?

This function identifies games with the strongest player ratings. I also require at least 1,000 reviews so that a game with only a few positive reviews does not automatically rank at the top.


```python
def highest_rated(games):
    reviewed = games[games["num_reviews_total"] >= 1000]

    rated = reviewed.sort_values(
        "pct_pos_total",
        ascending=False
    )

    return rated[["name", "pct_pos_total", "num_reviews_total"]].head(5)

highest_rated(games)
```


## Question 3: Additional Analysis Using SciPy

After identifying the highest-rated games, I used SciPy to see whether games with better player ratings also tend to have more players.

SciPy calculates the correlation between positive review percentages and peak concurrent players. This helps streamers understand whether popularity and ratings are closely related or should be considered separately when choosing a game.


```python
from scipy.stats import spearmanr

comparison = games[["peak_ccu", "pct_pos_total"]].dropna()

correlation, p_value = spearmanr(
    comparison["peak_ccu"],
    comparison["pct_pos_total"]
)

print("Popularity and ratings correlation:", round(correlation, 2))

if correlation > 0.3:
    print("More popular games tend to have higher ratings.")
elif correlation < -0.3:
    print("More popular games tend to have lower ratings.")
else:
    print("Popularity and ratings have a weak relationship.")
```


### What the Results Mean

The correlation was 0.02, showing almost no relationship between popularity and positive player ratings in this dataset.

This suggests that a game can be highly rated without having a large player base. StreamPick considers both factors separately when recommending games to streamers.


## SteamPick Recommendation System

This class combines popularity, average playtime, and player ratings to recommend games for streamers. Each factor receives equal weight in the final ranking.Each game receives a rank for popularity, playtime, and ratings. The ranks are added together, and the five games with the lowest combined scores become the recommendations.


```python
class SteamPick:
    def __init__(self, games):
        self.games = games.copy()

    def recommend(self):
        ranked = self.games.copy()

        ranked["popularity_rank"] = ranked["peak_ccu"].rank(
            ascending=False, pct=True
        )

        ranked["playtime_rank"] = ranked["avg_playtime_hours"].rank(
            ascending=False, pct=True
        )

        ranked["rating_rank"] = ranked["pct_pos_total"].rank(
            ascending=False, pct=True
        )

        ranked["score"] = (
            ranked["popularity_rank"]
            + ranked["playtime_rank"]
            + ranked["rating_rank"]
        )

        return ranked.sort_values("score").head(5)[
            ["name", "peak_ccu", "avg_playtime_hours",
             "pct_pos_total", "score"]
        ]

picker = SteamPick(games)
recommendations = picker.recommend()

print("Top 5 games recommended for streamers:")
display(recommendations)
```


## Final Recommendation

SteamPick recommends games by combining three factors: popularity, average playtime, and positive player ratings.

The final ranking gives each factor equal importance. This helps a streamer consider multiple characteristics rather than choosing a game based only on popularity.

The results are based on the Steam dataset and do not guarantee streaming success.
